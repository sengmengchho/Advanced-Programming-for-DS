# Topic 1: Exact Match (EM) evaluation.

Runs a small educational example with invented predictions (NOT real LLM outputs).
No third-party packages needed.

```python
def normalize(text: str) -> str:
    """Lowercase and collapse whitespace. Keep punctuation unchanged."""
    return " ".join(text.lower().split())


def exact_match(prediction: str, reference: str, normalized: bool = False) -> int:
    """Return 1 if strings match under chosen rules, else 0."""
    if normalized:
        return int(normalize(prediction) == normalize(reference))
    return int(prediction == reference)


def em_score(predictions: list[str], references: list[str], normalized: bool = False) -> float:
    """Return Exact Match percentage (0 to 100)."""
    if len(predictions) != len(references):
        raise ValueError("The number of predictions must equal the number of references.")
    if not references:
        raise ValueError("At least one question is required.")

    matches = sum(
        exact_match(pred, ref, normalized=normalized)
        for pred, ref in zip(predictions, references)
    )
    return matches / len(references) * 100


def main() -> None:
    questions = [
        "What is the capital of Cambodia?",
        "What is 2 + 2?",
        "What is the largest ocean on Earth?",
        "Which programming language uses .py files?",
        "In which country is Angkor Wat?",
    ]
    references = ["Phnom Penh", "4", "Pacific Ocean", "Python", "Cambodia"]

    # INVENTED EXAMPLES: replace with genuine output from actual LLMs for real experiments.
    predictions = {
        "Demo Model A": ["Phnom Penh", "4", "Pacific Ocean", "Python programming", "Cambodia"],
        "Demo Model B": ["phnom penh", "4", "Pacific Ocean", "Python", "Cambodia"],
    }

    print("Topic 1: Exact Match (EM) Evaluation")
    print("NOTE: Model A/B predictions are invented educational examples, NOT real LLM results.\n")
    print(f"{'Model':<16} {'Raw EM':>10} {'Normalized EM':>16}")
    print("-" * 44)
    for model_name, answers in predictions.items():
        raw = em_score(answers, references)
        norm = em_score(answers, references, normalized=True)
        print(f"{model_name:<16} {raw:>9.1f}% {norm:>15.1f}%")

    print("\nPer-question results for Demo Model A (raw comparison):")
    for i, (question, reference, answer) in enumerate(
        zip(questions, references, predictions["Demo Model A"]), start=1
    ):
        result = "MATCH" if exact_match(answer, reference) else "NO MATCH"
        print(f"{i}. {question}\n   Reference: {reference!r} | Answer: {answer!r} -> {result}")


if __name__ == "__main__":
    main()
```

# Topic 3: Precision, Recall, and F1-score.

All Demo Model predictions are INVENTED learning examples, not real LLM outputs.

```python
import csv
from pathlib import Path


def evaluate_binary(true_labels: list[str], predicted_labels: list[str]) -> dict:
    """Compute confusion matrix and metrics for the 'Positive' class."""
    if not true_labels or len(true_labels) != len(predicted_labels):
        raise ValueError("Provide non-empty true/predicted labels of equal length.")
    valid = {"Positive", "Negative"}
    if any(label not in valid for label in true_labels + predicted_labels):
        raise ValueError("Labels must be either 'Positive' or 'Negative'.")

    tp = sum(a == "Positive" and p == "Positive" for a, p in zip(true_labels, predicted_labels))
    tn = sum(a == "Negative" and p == "Negative" for a, p in zip(true_labels, predicted_labels))
    fp = sum(a == "Negative" and p == "Positive" for a, p in zip(true_labels, predicted_labels))
    fn = sum(a == "Positive" and p == "Negative" for a, p in zip(true_labels, predicted_labels))

    precision = tp / (tp + fp) if (tp + fp) else 0.0
    recall = tp / (tp + fn) if (tp + fn) else 0.0
    f1 = 2 * precision * recall / (precision + recall) if (precision + recall) else 0.0
    accuracy = (tp + tn) / len(true_labels)

    return {"tp": tp, "tn": tn, "fp": fp, "fn": fn,
            "precision": precision * 100, "recall": recall * 100,
            "f1": f1 * 100, "accuracy": accuracy * 100}


def main() -> None:
    print("Topic 3: Precision, Recall, and F1-score")
    print("IMPORTANT: Model predictions below are INVENTED examples, not real LLM outputs.\n")

    # Example 1: 8 TP, 2 FP, 2 FN (3 TN added to complete the dataset).
    example_1_true = ["Positive"] * 8 + ["Negative"] * 2 + ["Positive"] * 2 + ["Negative"] * 3
    example_1_pred = ["Positive"] * 8 + ["Positive"] * 2 + ["Negative"] * 2 + ["Negative"] * 3
    score_1 = evaluate_binary(example_1_true, example_1_pred)
    print(f"Example 1 (Spam): Precision={score_1['precision']:.2f}%, Recall={score_1['recall']:.2f}%, F1={score_1['f1']:.2f}%")

    # Example 2: 6 TP, 2 FP, 4 FN (3 TN added to complete the dataset).
    example_2_true = ["Positive"] * 6 + ["Negative"] * 2 + ["Positive"] * 4 + ["Negative"] * 3
    example_2_pred = ["Positive"] * 6 + ["Positive"] * 2 + ["Negative"] * 4 + ["Negative"] * 3
    score_2 = evaluate_binary(example_2_true, example_2_pred)
    print(f"Example 2 (Negative Reviews): Precision={score_2['precision']:.2f}%, Recall={score_2['recall']:.2f}%, F1={score_2['f1']:.2f}%\n")

    # Both demo models evaluate the SAME ten reference labels.
    true_labels = ["Positive"] * 5 + ["Negative"] * 5
    models = {
        # TP=4, TN=4, FP=1, FN=1
        "Demo Model A": ["Positive"] * 4 + ["Negative"] + ["Positive"] + ["Negative"] * 4,
        # TP=5, TN=3, FP=2, FN=0
        "Demo Model B": ["Positive"] * 5 + ["Positive"] * 2 + ["Negative"] * 3,
    }

    print(f"{'Model':<15} {'Precision':>10} {'Recall':>10} {'F1-score':>10} {'Accuracy':>10}")
    print("-" * 59)
    rows = []
    for name, predictions in models.items():
        score = evaluate_binary(true_labels, predictions)
        print(f"{name:<15} {score['precision']:>9.2f}% {score['recall']:>9.2f}% {score['f1']:>9.2f}% {score['accuracy']:>9.2f}%")
        rows.append({"model": name, **{k: round(v, 2) if isinstance(v, float) else v for k, v in score.items()},
                     "data_type": "INVENTED DEMO - NOT REAL LLM OUTPUT"})

    if "__file__" in globals():
        project_root = Path(__file__).resolve().parent.parent
    else:
        current_dir = Path.cwd()
        project_root = current_dir.parent if current_dir.name == "code" else current_dir
    output = project_root / "results/topic_03_precision_recall_f1_demo_comparison.csv"
    output.parent.mkdir(parents=True, exist_ok=True)
    with output.open("w", newline="", encoding="utf-8") as handle:
        writer = csv.DictWriter(handle, fieldnames=list(rows[0]))
        writer.writeheader()
        writer.writerows(rows)
    print(f"\nDemo results saved to {output}")
    print("For a real experiment, replace demo predictions with two real LLMs' outputs.")


if __name__ == "__main__":
    main()
```

# Topic 4: BLEU Score â€” beginner-friendly, standard-library implementation.

Measures corpus-level BLEU-1 and BLEU-4 with modified n-gram precision,
clipped counts, equal n-gram weights and a brevity penalty. No smoothing.

The included model predictions are INVENTED learning examples, not actual
outputs from LLMs. For real experiments, provide a CSV of real predictions.


```python
import argparse
import csv
import math
from collections import Counter
from pathlib import Path


if "__file__" in globals():
    PROJECT_ROOT = Path(__file__).resolve().parent.parent
else:
    current_dir = Path.cwd()
    PROJECT_ROOT = current_dir.parent if current_dir.name == "code" else current_dir
DEFAULT_INPUT = PROJECT_ROOT / "data" / "topic_04_bleu_examples.csv"
RESULTS = PROJECT_ROOT / "results"


def tokenize(text: str) -> list[str]:
    """Split on whitespace after lowercasing (simple educational tokenizer)."""
    return text.lower().split()


def ngram_counts(tokens: list[str], n: int) -> Counter:
    """Count consecutive n-word sequences."""
    return Counter(tuple(tokens[i:i + n]) for i in range(len(tokens) - n + 1))


def corpus_bleu(references: list[str], predictions: list[str], max_n: int = 4) -> dict:
    """Calculate unsmoothed, single-reference corpus BLEU for 1 <= n <= 4.

    BLEU = BP * exp(average(log(modified_precision_n))).
    If a modified precision is zero, unsmoothed BLEU is zero.
    """
    if not references or len(references) != len(predictions):
        raise ValueError("References and predictions must be non-empty and have equal lengths.")
    if max_n not in (1, 2, 3, 4):
        raise ValueError("max_n must be 1, 2, 3, or 4.")
    if any(not text.strip() for text in references):
        raise ValueError("Reference sentences cannot be empty.")

    clipped_counts = [0] * max_n
    total_counts = [0] * max_n
    reference_length = 0
    prediction_length = 0

    for reference, prediction in zip(references, predictions):
        ref_tokens = tokenize(reference)
        pred_tokens = tokenize(prediction)
        reference_length += len(ref_tokens)
        prediction_length += len(pred_tokens)

        for n in range(1, max_n + 1):
            ref_ngrams = ngram_counts(ref_tokens, n)
            pred_ngrams = ngram_counts(pred_tokens, n)
            clipped_counts[n - 1] += sum(
                min(count, ref_ngrams[gram]) for gram, count in pred_ngrams.items()
            )
            total_counts[n - 1] += sum(pred_ngrams.values())

    precisions = [
        matched / total if total else 0.0
        for matched, total in zip(clipped_counts, total_counts)
    ]
    if prediction_length == 0:
        bp = 0.0
    elif prediction_length >= reference_length:
        bp = 1.0
    else:
        bp = math.exp(1 - reference_length / prediction_length)

    if any(precision == 0 for precision in precisions):
        score = 0.0  # No smoothing is used in this simple demonstration.
    else:
        score = bp * math.exp(sum(math.log(p) for p in precisions) / max_n)

    return {
        "bleu": score * 100,
        "brevity_penalty": bp,
        "precisions": precisions,
        "reference_length": reference_length,
        "prediction_length": prediction_length,
    }
```

```python
def load_examples(file_path: Path) -> list[dict[str, str]]:
    if not file_path.is_file():
        raise FileNotFoundError(f"Input CSV does not exist: {file_path}")

    with file_path.open("r", encoding="utf-8-sig", newline="") as handle:
        reader = csv.DictReader(handle)
        columns = {"reference", "model_a", "model_b"}
        if not columns.issubset(reader.fieldnames or []):
            raise ValueError("CSV columns must include: reference, model_a, model_b")
        rows = list(reader)

    if not rows:
        raise ValueError("Input CSV must contain at least one example.")
    if any(not (row["reference"] or "").strip() for row in rows):
        raise ValueError("Every example needs a non-empty reference sentence.")
    if any(row["model_a"] is None or row["model_b"] is None for row in rows):
        raise ValueError("Every example must provide columns for both model predictions.")
    return rows


def save_csv(path: Path, rows: list[dict], columns: list[str]) -> None:
    path.parent.mkdir(parents=True, exist_ok=True)
    with path.open("w", encoding="utf-8", newline="") as handle:
        writer = csv.DictWriter(handle, fieldnames=columns)
        writer.writeheader()
        writer.writerows(rows)


def main() -> None:
    parser = argparse.ArgumentParser(description="Evaluate BLEU-1 and BLEU-4 for two models.")
    parser.add_argument("--input", type=Path, default=DEFAULT_INPUT,
                        help="CSV containing reference,model_a,model_b columns")
    args = parser.parse_args([] if "get_ipython" in globals() else None)

    print("Topic 4: BLEU Score")
    print("NOTE: The bundled data uses INVENTED model predictions, not real LLM outputs.\n")

    print("Example 1: An exact match")
    exact = corpus_bleu(["the cat is on the mat"], ["the cat is on the mat"], 4)
    print(f"BLEU-4 = {exact['bleu']:.2f}% (100% means a perfect n-gram match)\n")

    print("Example 2: A shorter prediction")
    shortened = corpus_bleu(["the cat is on the mat"], ["the cat is on mat"], 1)
    print(f"BLEU-1 = {shortened['bleu']:.2f}% (shorter answers receive a brevity penalty)\n")

    rows = load_examples(args.input)
    references = [row["reference"] for row in rows]
    models = {"Demo Model A": [row["model_a"] for row in rows],
              "Demo Model B": [row["model_b"] for row in rows]}

    print(f"Comparison using the SAME {len(rows)} references:\n")
    print(f"{'Model':<16}{'BLEU-1':>12}{'BLEU-4':>12}{'BP (4-gram)':>15}")
    print("-" * 55)
    comparison_rows = []
    for model_name, predictions in models.items():
        score_1 = corpus_bleu(references, predictions, max_n=1)
        score_4 = corpus_bleu(references, predictions, max_n=4)
        print(f"{model_name:<16}{score_1['bleu']:>11.2f}%{score_4['bleu']:>11.2f}%"
              f"{score_4['brevity_penalty']:>15.4f}")
        comparison_rows.append({
            "model": model_name,
            "number_of_sentences": len(rows),
            "bleu_1_percent": f"{score_1['bleu']:.2f}",
            "bleu_4_percent": f"{score_4['bleu']:.2f}",
            "brevity_penalty": f"{score_4['brevity_penalty']:.4f}",
            "reference_tokens": score_4["reference_length"],
            "prediction_tokens": score_4["prediction_length"],
            "note": "INVENTED DEMO - NOT REAL LLM OUTPUT",
        })

    comparison_path = RESULTS / "topic_04_bleu_demo_comparison.csv"
    save_csv(comparison_path, comparison_rows, list(comparison_rows[0]))
    print(f"\nComparison saved: {comparison_path}")
    print("To conduct a real experiment, replace the example model columns in a copy of")
    print("the input CSV with actual outputs from two LLMs and pass --input YOUR_FILE.csv")


if __name__ == "__main__":
    main()
```

# Topic 5: ROUGE Score — beginner-friendly, standard-library demonstration.

Computes ROUGE-1, ROUGE-2 and ROUGE-L precision, recall and F1 for
reference summaries and model-generated summaries. Per-document scores are
macro-averaged for model comparisons (not a corpus-pooled ROUGE calculation).

The bundled model predictions are INVENTED EXAMPLES, not real LLM outputs.


```python
import argparse
import csv
import re
from collections import Counter
from pathlib import Path

if "__file__" in globals():
    ROOT = Path(__file__).resolve().parent.parent
else:
    current_dir = Path.cwd()
    ROOT = current_dir.parent if current_dir.name == "code" else current_dir
DEFAULT_INPUT = ROOT / "data" / "topic_05_rouge_examples.csv"
RESULTS_DIR = ROOT / "results"


def tokenize(text: str) -> list[str]:
    """Lowercase, separate English-like words, and ignore punctuation."""
    return re.findall(r"\b\w+\b", text.casefold(), flags=re.UNICODE)


def ngram_counts(tokens: list[str], n: int) -> Counter:
    """Count neighboring sequences of n words."""
    return Counter(tuple(tokens[i:i + n]) for i in range(len(tokens) - n + 1))


def precision_recall_f1(overlap: int, predicted_total: int, reference_total: int) -> dict:
    """Return percentages; score is 0 if a denominator is zero."""
    precision = overlap / predicted_total if predicted_total else 0.0
    recall = overlap / reference_total if reference_total else 0.0
    f1 = 2 * precision * recall / (precision + recall) if precision + recall else 0.0
    return {"precision": 100 * precision, "recall": 100 * recall, "f1": 100 * f1}


def rouge_n(reference: str, prediction: str, n: int) -> dict:
    """ROUGE-N from clipped matching counts for one reference/prediction pair."""
    ref_grams = ngram_counts(tokenize(reference), n)
    pred_grams = ngram_counts(tokenize(prediction), n)
    overlap = sum(min(count, pred_grams[gram]) for gram, count in ref_grams.items())
    return precision_recall_f1(overlap, sum(pred_grams.values()), sum(ref_grams.values()))


def lcs_length(a: list[str], b: list[str]) -> int:
    """Length of longest common subsequence (words stay in order, not necessarily adjacent)."""
    previous = [0] * (len(b) + 1)
    for word_a in a:
        current = [0] * (len(b) + 1)
        for j, word_b in enumerate(b, 1):
            current[j] = previous[j - 1] + 1 if word_a == word_b else max(previous[j], current[j - 1])
        previous = current
    return previous[-1]


def rouge_l(reference: str, prediction: str) -> dict:
    """Simple sentence-level ROUGE-L based on word longest common subsequence."""
    ref_tokens = tokenize(reference)
    pred_tokens = tokenize(prediction)
    overlap = lcs_length(ref_tokens, pred_tokens)
    return precision_recall_f1(overlap, len(pred_tokens), len(ref_tokens))


def score_summary(reference: str, prediction: str) -> dict:
    if not reference.strip() or not tokenize(reference):
        raise ValueError("Reference summary must contain at least one word.")
    return {
        "rouge_1": rouge_n(reference, prediction, n=1),
        "rouge_2": rouge_n(reference, prediction, n=2),
        "rouge_l": rouge_l(reference, prediction),
    }


def load_examples(path: Path) -> list[dict[str, str]]:
    if not path.is_file():
        raise FileNotFoundError(f"Could not find the CSV file: {path}")
    with path.open("r", encoding="utf-8-sig", newline="") as file:
        reader = csv.DictReader(file)
        required = {"reference", "model_a", "model_b"}
        if not required.issubset(reader.fieldnames or []):
            raise ValueError("Input CSV needs columns: reference, model_a, model_b")
        rows = list(reader)
    if not rows:
        raise ValueError("Input CSV must contain at least one example.")
    for row in rows:
        if not row["reference"] or not tokenize(row["reference"]):
            raise ValueError("Every reference must contain at least one word.")
        if row["model_a"] is None or row["model_b"] is None:
            raise ValueError("Every row must contain both model predictions (may be empty).")
    return rows
```

```python
def save_csv(path: Path, rows: list[dict], headers: list[str]) -> None:
    path.parent.mkdir(parents=True, exist_ok=True)
    with path.open("w", encoding="utf-8", newline="") as file:
        writer = csv.DictWriter(file, fieldnames=headers)
        writer.writeheader()
        writer.writerows(rows)


def main() -> None:
    parser = argparse.ArgumentParser(description="Compare two models with ROUGE-1, ROUGE-2 and ROUGE-L")
    parser.add_argument("--input", type=Path, default=DEFAULT_INPUT,
                        help="CSV with reference, model_a, model_b columns")
    args, _ = parser.parse_known_args()

    print("Topic 5: ROUGE Score")
    print("NOTE: Built-in model outputs are INVENTED, not real LLM predictions.\n")

    example_1 = score_summary("the cat sat on the mat", "the cat sat on mat")
    print("Example 1 — shorter summary")
    print("ROUGE-1: Precision={precision:.2f}%, Recall={recall:.2f}%, F1={f1:.2f}%".format(
        **example_1["rouge_1"]))

    example_2 = score_summary("the cat eats fresh fish", "fresh fish the cat eats")
    print("\nExample 2 — same words, different order")
    for metric, label in [("rouge_1", "ROUGE-1"), ("rouge_2", "ROUGE-2"),
                          ("rouge_l", "ROUGE-L")]:
        print(f"{label}: F1={example_2[metric]['f1']:.2f}%")

    examples = load_examples(args.input)
    print(f"\nModel comparison using the SAME {len(examples)} reference summaries:")
    print(f"{'Model':<17}{'ROUGE-1 F1':>13}{'ROUGE-2 F1':>13}{'ROUGE-L F1':>13}")
    print("-" * 56)
    comparison_rows = []
    detailed_rows = []

    for model_col, model_name in [("model_a", "Demo Model A"), ("model_b", "Demo Model B")]:
        all_scores = []
        for index, row in enumerate(examples, 1):
            metrics = score_summary(row["reference"], row[model_col])
            all_scores.append(metrics)
            details = {
                "example_id": index,
                "model": model_name,
                "reference": row["reference"],
                "prediction": row[model_col],
            }
            for metric in ("rouge_1", "rouge_2", "rouge_l"):
                for value in ("precision", "recall", "f1"):
                    details[f"{metric}_{value}_percent"] = f"{metrics[metric][value]:.2f}"
            detailed_rows.append(details)

        averages = {
            f"{metric}_{value}_percent": sum(s[metric][value] for s in all_scores) / len(all_scores)
            for metric in ("rouge_1", "rouge_2", "rouge_l")
            for value in ("precision", "recall", "f1")
        }
        print(f"{model_name:<17}{averages['rouge_1_f1_percent']:>12.2f}%"
              f"{averages['rouge_2_f1_percent']:>12.2f}%"
              f"{averages['rouge_l_f1_percent']:>12.2f}%")
        comparison_rows.append({
            "model": model_name,
            "number_of_summaries": len(examples),
            **{key: f"{value:.2f}" for key, value in averages.items()},
            "aggregation": "macro mean of per-summary scores",
            "note": "INVENTED DEMO - NOT REAL LLM OUTPUT",
        })

    comparison_file = RESULTS_DIR / "topic_05_rouge_demo_comparison.csv"
    details_file = RESULTS_DIR / "topic_05_rouge_demo_predictions.csv"
    save_csv(comparison_file, comparison_rows, list(comparison_rows[0]))
    save_csv(details_file, detailed_rows, list(detailed_rows[0]))
    print(f"\nSaved comparison: {comparison_file}")
    print(f"Saved prediction details: {details_file}")
    print("For real experiments, use the same references for two actual LLMs and")
    print("pass their generated summaries in your own CSV with --input MY_DATA.csv")


if __name__ == "__main__":
    main()
```

# Topic 6: METEOR Score — beginner-friendly NLTK demonstration.

Measures sentence-level METEOR for generated text against references, then
reports the arithmetic mean of per-sentence scores for two demo models.

Default: NLTK's METEOR formula, stemming and a LIMITED offline synonym list.
Use --full-wordnet to enable NLTK's complete WordNet synonym matching.

The bundled predictions are INVENTED, not real LLM outputs.

```python
import argparse
import csv
import re
import sys
from pathlib import Path
from statistics import mean

from nltk.stem import PorterStemmer
from nltk.translate.meteor_score import meteor_score

if "__file__" in globals():
    ROOT = Path(__file__).resolve().parent.parent
else:
    current_dir = Path.cwd()
    ROOT = current_dir.parent if current_dir.name == "code" else current_dir
DATA_FILE = ROOT / "data" / "topic_06_meteor_examples.csv"
RESULTS_DIR = ROOT / "results"


def tokenize(text: str) -> list[str]:
    """Simple case-insensitive English word tokenization."""
    return re.findall(r"[a-z]+(?:'[a-z]+)?", text.casefold())


class _DemoLemma:
    def __init__(self, word: str):
        self._word = word

    def name(self) -> str:
        return self._word


class _DemoSynset:
    def __init__(self, words: set[str]):
        self._words = words

    def lemmas(self) -> list[_DemoLemma]:
        return [_DemoLemma(word) for word in sorted(self._words)]


class SmallOfflineWordNet:
    """A tiny synonym dictionary for reproducible OFFLINE classroom examples.

    NOT the real WordNet corpus. METEOR still uses NLTK's exact/stem matching,
    precision, recall, and fragmentation-penalty formula. Only the available
    synonym vocabulary is restricted. Use --full-wordnet for genuine WordNet.
    """

    def __init__(self) -> None:
        stemmer = PorterStemmer()
        groups = (
            ("car", "automobile"),
            ("fast", "quick"),
            ("big", "large"),
            ("kid", "child", "children", "kids"),
            ("home", "house"),
            ("begin", "start"),
            ("buy", "purchase"),
            ("happy", "glad"),
            ("smart", "clever"),
        )
        self.synonym_map = {}
        for group in groups:
            forms = {stemmer.stem(word) for word in group}
            for form in forms:
                self.synonym_map[form] = forms

    def synsets(self, word: str) -> list[_DemoSynset]:
        words = self.synonym_map.get(word)
        return [_DemoSynset(words)] if words else []

```

```python
def scorer_wordnet(use_full_wordnet: bool):
    if not use_full_wordnet:
        return SmallOfflineWordNet()

    from nltk.corpus import wordnet
    try:
        wordnet.ensure_loaded()
    except LookupError as exc:
        raise RuntimeError(
            "Full WordNet data is missing. Run: python -m nltk.downloader wordnet omw-1.4\n"
            "Or run without --full-wordnet to use the small offline demonstration dictionary."
        ) from exc
    return wordnet


def score_sentence(reference: str, prediction: str, wordnet) -> float:
    """Compute single-reference, sentence-level METEOR, scaled to a percentage."""
    reference_words = tokenize(reference)
    if not reference_words:
        raise ValueError("Reference text must contain at least one English word.")
    predicted_words = tokenize(prediction)
    if not predicted_words:
        return 0.0
    score = meteor_score([reference_words], predicted_words, wordnet=wordnet)
    return score * 100.0


def read_examples(path: Path) -> list[dict[str, str]]:
    if not path.is_file():
        raise FileNotFoundError(f"No CSV file found: {path}")
    with path.open("r", encoding="utf-8-sig", newline="") as fh:
        reader = csv.DictReader(fh)
        required = {"reference", "model_a", "model_b"}
        if not required.issubset(reader.fieldnames or []):
            raise ValueError("CSV must include columns: reference,model_a,model_b")
        rows = list(reader)
    if not rows:
        raise ValueError("CSV must contain at least one example.")
    for index, row in enumerate(rows, start=1):
        if not tokenize(row["reference"] or ""):
            raise ValueError(f"Row {index}: reference needs at least one English word.")
        if row["model_a"] is None or row["model_b"] is None:
            raise ValueError(f"Row {index}: model predictions cannot be missing.")
    return rows


def write_csv(path: Path, rows: list[dict]) -> None:
    path.parent.mkdir(exist_ok=True, parents=True)
    with path.open("w", newline="", encoding="utf-8") as fh:
        writer = csv.DictWriter(fh, fieldnames=list(rows[0]))
        writer.writeheader()
        writer.writerows(rows)
```

```python
def main() -> None:
    parser = argparse.ArgumentParser(description="Topic 6: METEOR evaluation")
    parser.add_argument("--input", type=Path, default=DATA_FILE,
                        help="CSV file with reference,model_a,model_b columns")
    parser.add_argument("--full-wordnet", action="store_true",
                        help="Use the real NLTK WordNet corpus instead of a small offline synonym dictionary")
    args = parser.parse_args([] if "get_ipython" in globals() else None)
    try:
        wordnet = scorer_wordnet(args.full_wordnet)
    except RuntimeError as error:
        print(f"SETUP NEEDED: {error}", file=sys.stderr)
        raise SystemExit(2) from None
    mode = "complete NLTK WordNet" if args.full_wordnet else "limited offline synonym dictionary"

    print("Topic 6: METEOR Score")
    print(f"Synonym matching: {mode}")
    print("Bundled data are INVENTED examples, NOT real LLM predictions.\n")

    print("Example 1 — identical sentences:")
    example1 = score_sentence("the student reads a book", "the student reads a book", wordnet)
    print(f"METEOR: {example1:.2f}%")
    print("Example 2 — matching synonyms:")
    example2 = score_sentence("the car is fast", "the automobile is quick", wordnet)
    print(f"METEOR: {example2:.2f}%")
    reordered = score_sentence("the car is fast", "quick is the automobile", wordnet)
    print(f"Same matched words but different order: {reordered:.2f}%")

    rows = read_examples(args.input)
    print(f"\nComparison on the same {len(rows)} reference sentences:")
    print(f"{'Model':<18}{'Average METEOR':>16}")
    print("-" * 34)

    comparisons = []
    details = []
    for column, model in (("model_a", "Demo Model A"), ("model_b", "Demo Model B")):
        scores = []
        for index, row in enumerate(rows, start=1):
            score = score_sentence(row["reference"], row[column], wordnet)
            scores.append(score)
            details.append({
                "example_id": index,
                "model": model,
                "reference": row["reference"],
                "prediction": row[column],
                "meteor_percent": f"{score:.2f}",
            })
        average = mean(scores)
        print(f"{model:<18}{average:>15.2f}%")
        comparisons.append({
            "model": model,
            "number_of_sentences": len(rows),
            "average_meteor_percent": f"{average:.2f}",
            "aggregation": "arithmetic mean of sentence METEOR percentages",
            "synonym_mode": mode,
            "note": "INVENTED DEMO - NOT REAL LLM OUTPUTS",
        })

    suffix = "wordnet" if args.full_wordnet else "offline"
    summary_path = RESULTS_DIR / f"topic_06_meteor_demo_comparison_{suffix}.csv"
    detail_path = RESULTS_DIR / f"topic_06_meteor_demo_predictions_{suffix}.csv"
    write_csv(summary_path, list(comparisons[0]), comparisons)
    write_csv(detail_path, list(details[0]), details)
    print(f"\nSaved summary: {summary_path.relative_to(ROOT)}")
    print(f"Saved detailed results: {detail_path.relative_to(ROOT)}")
    if not args.full_wordnet:
        print("NOTE: Synonym matching is limited; use --full-wordnet for broader coverage.")


if __name__ == "__main__":
    main()
```

# Topic 7: BERTScore for LLM evaluation.

Run offline educational illustration (NOT a real BERTScore measurement):
    python code/topic_07_bertscore.py


For a genuine LLM-vs-LLM experiment, replace the two demo-model columns in
 data/topic_07_bertscore_examples.csv with actual outputs from two LLMs.


```python
import argparse
import csv
import json
import sys
from pathlib import Path

if "__file__" in globals():
    ROOT = Path(__file__).resolve().parents[1]
else:
    current_dir = Path.cwd()
    ROOT = current_dir.parent if current_dir.name == "code" else current_dir
INPUT = ROOT / "data" / "topic_07_bertscore_examples.csv"
TOY_INPUT = ROOT / "data" / "topic_07_toy_similarities.json"
OUT_DIR = ROOT / "results"


def harmonic_f1(precision: float, recall: float) -> float:
    return 2 * precision * recall / (precision + recall) if precision + recall else 0.0


def scores_from_similarity_matrix(matrix: list[list[float]]) -> tuple[float, float, float]:
    """Educational implementation of BERTScore's token-matching equations.

    Matrix rows = reference tokens; columns = candidate tokens.
    All similarities here are ASSUMED, not obtained from a language model.
    Real BERTScore obtains the values from contextual token embeddings.
    """
    if not matrix or not matrix[0]:
        raise ValueError("Matrix must have at least one row and column")
    rows, cols = len(matrix), len(matrix[0])
    if any(len(row) != cols for row in matrix):
        raise ValueError("Matrix must be rectangular")
    if any(not (-1 <= v <= 1) for row in matrix for v in row):
        raise ValueError("Similarities should be between -1 and 1")
    recall = sum(max(row) for row in matrix) / rows
    precision = sum(max(matrix[i][j] for i in range(rows)) for j in range(cols)) / cols
    return precision, recall, harmonic_f1(precision, recall)


def load_examples() -> list[dict[str, str]]:
    with INPUT.open(newline="", encoding="utf-8") as handle:
        rows = list(csv.DictReader(handle))
    if not rows or any(not row[field].strip() for row in rows for field in
                       ("id", "reference", "demo_model_a", "demo_model_b")):
        raise ValueError("Example CSV has missing data")
    return rows


def write_csv(path: Path, headers: list[str], rows: list[dict]) -> None:
    path.parent.mkdir(parents=True, exist_ok=True)
    with path.open("w", newline="", encoding="utf-8") as handle:
        writer = csv.DictWriter(handle, fieldnames=headers)
        writer.writeheader()
        writer.writerows(rows)


def save_results(examples, metrics, mode):
    """Store per-example and aggregated scores with an explicit provenance label."""
    per_rows=[]
    summary=[]
    label = ("TOY manually assigned similarities — NOT real BERTScore" if mode == "toy"
             else "Actual BERTScore applied to INVENTED model predictions")
    for model in ("demo_model_a", "demo_model_b"):
        values=[]
        for example, (p,r,f1) in zip(examples,metrics[model]):
            per_rows.append({"id":example["id"],"model":model,"reference":example["reference"],
                "prediction":example[model],"precision":f"{p:.6f}","recall":f"{r:.6f}",
                "f1":f"{f1:.6f}","score_origin":label})
            values.append((p,r,f1))
        n=len(values)
        summary.append({"model":model,"examples":n,
            "mean_precision":f"{sum(v[0] for v in values)/n:.6f}",
            "mean_recall":f"{sum(v[1] for v in values)/n:.6f}",
            "mean_f1":f"{sum(v[2] for v in values)/n:.6f}","score_origin":label})
    suffix="toy_demo" if mode=="toy" else "real_metric_on_demo_text"
    per_path=OUT_DIR/f"topic_07_bertscore_{suffix}_predictions.csv"
    summary_path=OUT_DIR/f"topic_07_bertscore_{suffix}_comparison.csv"
    write_csv(per_path,list(per_rows[0]),per_rows)
    write_csv(summary_path,list(summary[0]),summary)
    return summary, per_path, summary_path
```

```python
def run_toy(examples):
    """Completely offline illustration of mathematical matching, not BERTScore."""
    with TOY_INPUT.open(encoding="utf-8") as handle:
        items={case["id"]:case for case in json.load(handle)["cases"]}
    metrics={"demo_model_a":[],"demo_model_b":[]}
    for row in examples:
        example_id=row["id"]
        if example_id not in items:
            raise ValueError(f"Missing toy matrix for sample {example_id}")
        for model in metrics:
            matrix=items[example_id][model]
            if len(matrix)!=len(row["reference"].split()) or any(
                len(line)!=len(row[model].split()) for line in matrix
            ):
                raise ValueError(f"Toy token matrix shape mismatch for {example_id} / {model}")
            metrics[model].append(scores_from_similarity_matrix(matrix))
    return metrics


def run_real(examples, model_type: str):
    """Run official BERTScore. Requires installed dependencies and model weights."""
    try:
        from bert_score import score as bert_score
    except ImportError as exc:
        raise RuntimeError(
            "Install BERTScore first: python -m pip install -r requirements-topic-07.txt"
        ) from exc
    reference=[item["reference"] for item in examples]
    candidates=[row["demo_model_a"] for row in examples]+[row["demo_model_b"] for row in examples]
    references=reference+reference
    # Specifying the same model/settings across candidates ensures a fair comparison.
    # The first --real run may download pretrained weights from Hugging Face.
    p,r,f=bert_score(candidates,references,model_type=model_type,lang="en",idf=False,
                     rescale_with_baseline=False,batch_size=4,verbose=False)
    triples=list(zip(p.tolist(),r.tolist(),f.tolist()))
    n=len(examples)
    return {"demo_model_a":triples[:n],"demo_model_b":triples[n:]}


def main() -> int:
    parser=argparse.ArgumentParser(description=__doc__)
    parser.add_argument("--real",action="store_true",help="Run actual pretrained BERTScore instead of toy example")
    parser.add_argument("--model-type",default="distilbert-base-uncased",
                        help="Hugging Face model for actual BERTScore (default: distilbert-base-uncased)")
    args, _=parser.parse_known_args()
    examples=load_examples()
    try:
        metrics=run_real(examples,args.model_type) if args.real else run_toy(examples)
    except Exception as exc:
        print(f"Could not compute scores: {exc}",file=sys.stderr)
        return 1
    mode="real" if args.real else "toy"
    summary,per_path,summary_path=save_results(examples,metrics,mode)
    print("Topic 7: BERTScore")
    if not args.real:
        print("IMPORTANT: These are hand-assigned token similarities, NOT true BERTScore.")
    else:
        print(f"Actual BERTScore computed using: {args.model_type}")
        print("Text labeled Demo Model A/B is invented, not generated by actual LLMs.")
    print(f"{'Model':<18} {'Mean Precision':>15} {'Mean Recall':>15} {'Mean F1':>12}")
    print("-"*64)
    for result in summary:
        print(f"{result['model']:<18} {float(result['mean_precision'])*100:>14.2f}% "
              f"{float(result['mean_recall'])*100:>14.2f}% "
              f"{float(result['mean_f1'])*100:>11.2f}%")
    print(f"Saved: {per_path.relative_to(ROOT)}")
    print(f"Saved: {summary_path.relative_to(ROOT)}")
    return 0
```

# Topic 8: Word Error Rate (WER) evaluation.

WER counts word substitutions (S), deletions (D), and insertions (I):
    WER = (S + D + I) / number_of_reference_words * 100

The provided predictions are INVENTED DEMONSTRATIONS, not actual LLM outputs.
Uses Python's standard library only; no pip install needed.

```python
import argparse
import csv
import re
from pathlib import Path

if "__file__" in globals():
    ROOT = Path(__file__).resolve().parent.parent
else:
    current_dir = Path.cwd()
    ROOT = current_dir.parent if current_dir.name == "code" else current_dir
DEFAULT_INPUT = ROOT / "data" / "topic_08_wer_examples.csv"
DEFAULT_RESULTS_DIR = ROOT / "results"


def tokenize(text: str) -> list[str]:
    """Basic case-insensitive English-oriented tokenization.

    Keeps letters, digits and internal apostrophes; removes other punctuation.
    Word segmentation of scripts such as Khmer needs a proper tokenizer.
    """
    return re.findall(r"[^\W_]+(?:['’][^\W_]+)*", text.casefold(), re.UNICODE)


def count_word_errors(reference: str, prediction: str) -> dict:
    """Return a minimum-edit alignment with its S, D, I counts and WER.

    Ties are resolved consistently in the order: substitution, deletion, insertion.
    Counts for S/D/I may depend on tie-breaking, but total errors / WER do not.
    """
    ref = tokenize(reference)
    pred = tokenize(prediction)
    n, m = len(ref), len(pred)
    if n == 0:
        raise ValueError("The reference must contain at least one word for WER.")

    distance = [[0] * (m + 1) for _ in range(n + 1)]
    for i in range(n + 1):
        distance[i][0] = i
    for j in range(m + 1):
        distance[0][j] = j

    for i in range(1, n + 1):
        for j in range(1, m + 1):
            cost = 0 if ref[i - 1] == pred[j - 1] else 1
            distance[i][j] = min(
                distance[i - 1][j - 1] + cost,   # match or substitute
                distance[i - 1][j] + 1,          # delete reference word
                distance[i][j - 1] + 1,          # insert predicted word
            )

    substitutions = deletions = insertions = 0
    i, j = n, m
    while i > 0 or j > 0:
        if i > 0 and j > 0 and ref[i - 1] == pred[j - 1] and distance[i][j] == distance[i - 1][j - 1]:
            i -= 1
            j -= 1
        elif i > 0 and j > 0 and distance[i][j] == distance[i - 1][j - 1] + 1:
            substitutions += 1
            i -= 1
            j -= 1
        elif i > 0 and distance[i][j] == distance[i - 1][j] + 1:
            deletions += 1
            i -= 1
        elif j > 0 and distance[i][j] == distance[i][j - 1] + 1:
            insertions += 1
            j -= 1
        else:
            raise AssertionError("Unable to trace the optimal alignment.")

    total_errors = substitutions + deletions + insertions
    assert total_errors == distance[n][m]
    return {
        "reference_words": n,
        "predicted_words": m,
        "substitutions": substitutions,
        "deletions": deletions,
        "insertions": insertions,
        "total_errors": total_errors,
        "wer_percent": 100.0 * total_errors / n,
    }
```

```python
def load_examples(path: Path) -> list[dict]:
    with path.open("r", encoding="utf-8-sig", newline="") as file:
        reader = csv.DictReader(file)
        required = {"reference", "model_a", "model_b"}
        if not required.issubset(reader.fieldnames or []):
            raise ValueError("CSV needs columns: reference, model_a, model_b")
        rows = list(reader)
    if not rows:
        raise ValueError("CSV needs at least one example.")
    for index, row in enumerate(rows, 1):
        if not row.get("reference") or not tokenize(row["reference"]):
            raise ValueError(f"Row {index}: reference must contain at least one word.")
        if row.get("model_a") is None or row.get("model_b") is None:
            raise ValueError(f"Row {index}: both model columns are required (empty predictions allowed).")
    return rows


def save_csv(path: Path, records: list[dict]) -> None:
    path.parent.mkdir(parents=True, exist_ok=True)
    with path.open("w", encoding="utf-8", newline="") as file:
        writer = csv.DictWriter(file, fieldnames=list(records[0]))
        writer.writeheader()
        writer.writerows(records)


def run(input_file: Path, results_dir: Path) -> None:
    print("Topic 8: Word Error Rate (WER)")
    print("NOTE: Demo Model A/B are INVENTED text, not actual LLM outputs.\n")

    for label, reference, prediction in [
        ("Example 1 — substitution", "the cat is sleeping", "the dog is sleeping"),
        ("Example 2 — deletion and insertion", "the teacher explained a problem clearly", "yesterday the teacher explained a problem"),
    ]:
        score = count_word_errors(reference, prediction)
        print(f"{label}: S={score['substitutions']}, D={score['deletions']}, I={score['insertions']}, "
              f"N={score['reference_words']}, WER={score['wer_percent']:.2f}%")

    samples = load_examples(input_file)
    details = []
    comparisons = []
    print(f"\nModel comparison using the SAME {len(samples)} references:")
    print(f"{'Model':<17}{'S':>5}{'D':>5}{'I':>5}{'Errors':>9}{'Ref words':>12}{'WER':>10}")
    print("-" * 63)

    for model_field, model_name in (("model_a", "Demo Model A"), ("model_b", "Demo Model B")):
        total = {key: 0 for key in ("reference_words", "substitutions", "deletions", "insertions", "total_errors")}
        for idx, sample in enumerate(samples, 1):
            score = count_word_errors(sample["reference"], sample[model_field])
            for key in total:
                total[key] += score[key]
            details.append({
                "example_id": sample.get("example_id") or idx,
                "model": model_name,
                "reference": sample["reference"],
                "prediction": sample[model_field],
                **{key: (f"{value:.2f}" if key == "wer_percent" else value)
                   for key, value in score.items()},
                "note": "INVENTED DEMO - NOT REAL LLM OUTPUT",
            })
        corpus_wer = 100.0 * total["total_errors"] / total["reference_words"]
        print(f"{model_name:<17}{total['substitutions']:>5}{total['deletions']:>5}"
              f"{total['insertions']:>5}{total['total_errors']:>9}"
              f"{total['reference_words']:>12}{corpus_wer:>9.2f}%")
        comparisons.append({
            "model": model_name,
            "total_sentences": len(samples),
            **total,
            "corpus_wer_percent": f"{corpus_wer:.2f}",
            "aggregation": "sum errors / sum reference words (micro/corpus WER)",
            "note": "INVENTED DEMO - NOT REAL LLM OUTPUT",
        })

    comparisons_file = results_dir / "topic_08_wer_demo_comparison.csv"
    predictions_file = results_dir / "topic_08_wer_demo_predictions.csv"
    save_csv(comparisons_file, comparisons)
    save_csv(predictions_file, details)
    print(f"\nSaved comparison: {comparisons_file}")
    print(f"Saved detailed predictions: {predictions_file}")
    print("Lower WER is better. A score above 100% is possible with many insertions.")
    print("To evaluate real LLMs, replace example predictions with actual responses from two models.")


def main() -> None:
    parser = argparse.ArgumentParser(description="Compute word substitutions, deletions, insertions and corpus WER")
    parser.add_argument("--input", type=Path, default=DEFAULT_INPUT,
                        help="CSV containing reference, model_a, model_b")
    parser.add_argument("--results-dir", type=Path, default=DEFAULT_RESULTS_DIR,
                        help="Where to save the comparison and detailed result CSVs")
    arguments, _ = parser.parse_known_args()
    run(arguments.input, arguments.results_dir)


if __name__ == "__main__":
    main()
```

# Topic 9: Character Error Rate (CER).

CER = (substitutions + deletions + insertions) / reference characters * 100

The included Demo Model A/B predictions are invented learning examples,
NOT outputs collected from real LLMs.

```python
import argparse
import csv
import unicodedata
from pathlib import Path

if "__file__" in globals():
    ROOT = Path(__file__).resolve().parent.parent
else:
    current_dir = Path.cwd()
    ROOT = current_dir.parent if current_dir.name == "code" else current_dir
DEFAULT_INPUT = ROOT / "data" / "topic_09_cer_examples.csv"
DEFAULT_RESULTS_DIR = ROOT / "results"


def normalize_text(text: str) -> str:
    """Normalize Unicode (NFC), fold case, and trim outside whitespace.

    Spaces and punctuation *inside* the text ARE counted as characters.
    Counts are Unicode code points, not visual grapheme clusters.
    """
    return unicodedata.normalize("NFC", unicodedata.normalize("NFC", text.strip()).casefold())


def count_character_errors(reference: str, prediction: str) -> dict:
    """Calculate minimum character edits plus one deterministic S/D/I alignment.

    Ties use substitution, then deletion, then insertion. An equally optimal
    alignment may have different individual S/D/I counts, but the total edit
    distance and CER are unchanged.
    """
    ref = list(normalize_text(reference))
    pred = list(normalize_text(prediction))
    n, m = len(ref), len(pred)
    if n == 0:
        raise ValueError("Reference must contain at least one character after normalization.")

    dp = [[0] * (m + 1) for _ in range(n + 1)]
    for i in range(n + 1):
        dp[i][0] = i
    for j in range(m + 1):
        dp[0][j] = j

    for i in range(1, n + 1):
        for j in range(1, m + 1):
            cost = 0 if ref[i - 1] == pred[j - 1] else 1
            dp[i][j] = min(
                dp[i - 1][j - 1] + cost,  # match or substitution
                dp[i - 1][j] + 1,         # deletion from reference
                dp[i][j - 1] + 1,         # insertion into prediction
            )

    subs = dels = ins = 0
    i, j = n, m
    while i or j:
        if i and j and ref[i - 1] == pred[j - 1] and dp[i][j] == dp[i - 1][j - 1]:
            i -= 1
            j -= 1
        elif i and j and dp[i][j] == dp[i - 1][j - 1] + 1:
            subs += 1
            i -= 1
            j -= 1
        elif i and dp[i][j] == dp[i - 1][j] + 1:
            dels += 1
            i -= 1
        elif j and dp[i][j] == dp[i][j - 1] + 1:
            ins += 1
            j -= 1
        else:
            raise AssertionError("Could not trace minimum edit alignment.")

    errors = subs + dels + ins
    assert errors == dp[n][m]
    return {
        "reference_characters": n,
        "predicted_characters": m,
        "substitutions": subs,
        "deletions": dels,
        "insertions": ins,
        "total_errors": errors,
        "cer_percent": 100.0 * errors / n,
    }
```

```python
def load_examples(path: Path) -> list[dict]:
    with path.open("r", encoding="utf-8-sig", newline="") as file:
        reader = csv.DictReader(file)
        if not {"reference", "model_a", "model_b"}.issubset(reader.fieldnames or []):
            raise ValueError("CSV needs columns: reference,model_a,model_b")
        records = list(reader)
    if not records:
        raise ValueError("CSV must contain at least one row.")
    for index, record in enumerate(records, 1):
        if record.get("reference") is None or not normalize_text(record["reference"]):
            raise ValueError(f"Row {index}: reference must not be empty.")
        if record.get("model_a") is None or record.get("model_b") is None:
            raise ValueError(f"Row {index}: predictions must be present (empty text is allowed).")
    return records


def save_csv(path: Path, records: list[dict]) -> None:
    path.parent.mkdir(parents=True, exist_ok=True)
    with path.open("w", encoding="utf-8", newline="") as file:
        writer = csv.DictWriter(file, fieldnames=list(records[0].keys()))
        writer.writeheader()
        writer.writerows(records)


def run(input_file: Path, results_dir: Path) -> None:
    is_demo = input_file.resolve() == DEFAULT_INPUT.resolve()
    print("Topic 9: Character Error Rate (CER)")
    print("Lower CER is better. Character counts are Unicode code points.")
    if is_demo:
        print("Demo Model A and B are INVENTED text, NOT actual LLM outputs.\n")
    else:
        print("Custom predictions: verify their provenance before reporting them as real LLM outputs.\n")

    for name, reference, prediction in [
        ("Example 1 — substitution", "cat", "cut"),
        ("Example 2a — deletion", "hello", "helo"),
        ("Example 2b — insertion", "book", "books"),
    ]:
        score = count_character_errors(reference, prediction)
        print(
            f"{name}: S={score['substitutions']}, D={score['deletions']}, "
            f"I={score['insertions']}, N={score['reference_characters']}, "
            f"CER={score['cer_percent']:.2f}%"
        )

    samples = load_examples(input_file)
    detailed = []
    comparisons = []
    print(f"\nComparison on the SAME {len(samples)} references:")
    print(f"{'Model':<16}{'S':>5}{'D':>5}{'I':>5}{'Errors':>9}{'Ref chars':>12}{'CER':>10}")
    print("-" * 62)

    provenance = "INVENTED DEMO - NOT REAL LLM OUTPUT" if is_demo else "USER-PROVIDED - VERIFY SOURCE"
    for column, name in (("model_a", "Demo Model A"), ("model_b", "Demo Model B")):
        totals = {key: 0 for key in (
            "reference_characters", "substitutions", "deletions", "insertions", "total_errors"
        )}
        for index, sample in enumerate(samples, 1):
            score = count_character_errors(sample["reference"], sample[column])
            for key in totals:
                totals[key] += score[key]
            detailed.append({
                "example_id": sample.get("example_id") or index,
                "model": name,
                "reference": sample["reference"],
                "prediction": sample[column],
                **{key: (f"{value:.2f}" if key == "cer_percent" else value)
                   for key, value in score.items()},
                "note": provenance,
            })
        corpus_cer = 100.0 * totals["total_errors"] / totals["reference_characters"]
        print(
            f"{name:<16}{totals['substitutions']:>5}{totals['deletions']:>5}"
            f"{totals['insertions']:>5}{totals['total_errors']:>9}"
            f"{totals['reference_characters']:>12}{corpus_cer:>9.2f}%"
        )
        comparisons.append({
            "model": name,
            "total_examples": len(samples),
            **totals,
            "corpus_cer_percent": f"{corpus_cer:.2f}",
            "aggregation": "sum edits / sum reference characters (corpus CER)",
            "normalization": "NFC + casefold + strip outer whitespace; internal spaces/punctuation retained",
            "note": provenance,
        })

    suffix = "demo" if is_demo else "custom"
    comparison_path = results_dir / f"topic_09_cer_{suffix}_comparison.csv"
    predictions_path = results_dir / f"topic_09_cer_{suffix}_predictions.csv"
    save_csv(comparison_path, comparisons)
    save_csv(predictions_path, detailed)
    print(f"\nSaved comparison: {comparison_path}")
    print(f"Saved detailed predictions: {predictions_path}")
    print("CER can exceed 100% with enough insertions.")
    print("For a real experiment, replace the invented texts with actual responses from two models.")
```

```python
def main() -> None:
    parser = argparse.ArgumentParser(description="Count edit operations and calculate corpus CER")
    parser.add_argument("--input", type=Path, default=DEFAULT_INPUT,
                        help="CSV file with reference,model_a,model_b columns")
    parser.add_argument("--results-dir", type=Path, default=DEFAULT_RESULTS_DIR,
                        help="Folder in which result CSVs are saved")
    args, _ = parser.parse_known_args()
    run(args.input, args.results_dir)


if __name__ == "__main__":
    main()
```

# Topic 10: Compare LLM evaluation metrics across appropriate NLP tasks.

This script reuses the calculators already written for Topics 1, 4, 5, 6,
8 and 9. It compares two sets of INVENTED example predictions per task:
 - QA: Exact Match
 - Classification: Accuracy, Precision, Recall, F1
 - Text generation: BLEU-4, ROUGE-1, ROUGE-2, ROUGE-L, METEOR, BERTScore
 - Transcription: Word Error Rate (WER), Character Error Rate (CER)

A metric should be applied to a suitable task: these percentages are NOT a
single ranking across unrelated tasks or comparable numerical scales.

Setup, from repository root:
    python -m pip install -r requirements.txt
    python code/topic_10_comparative_evaluation.py

Optional REAL BERTScore (requires package + pretrained model download):
    python -m pip install -r requirements-topic-07.txt
    python code/topic_10_comparative_evaluation.py --real-bertscore

For genuine two-LLM experiments, replace columns model_a / model_b in COPIES
of the four CSVs in data/ with responses from two actual LLMs, then run:
    python code/topic_10_comparative_evaluation.py --data-dir my_real_data

```python
from __future__ import annotations

import argparse
import csv
import sys
from pathlib import Path
from statistics import mean

if "__file__" in globals():
    ROOT = Path(__file__).resolve().parents[1]
else:
    current_dir = Path.cwd()
    ROOT = current_dir.parent if current_dir.name == "code" else current_dir
CODE = ROOT / "code"
if str(CODE) not in sys.path:
    sys.path.insert(0, str(CODE))

# Metric functions are defined by the preceding notebook topic cells.

DATA_FILENAMES = {
    "question answering": "topic_10_qa_demo.csv",
    "classification": "topic_10_classification_demo.csv",
    "text generation": "topic_10_generation_demo.csv",
    "transcription": "topic_10_transcription_demo.csv",
}


def read_data(data_dir: Path) -> dict[str, list[dict[str, str]]]:
    """Read four tasks with identical columns; reject invalid input."""
    tables = {}
    for task, name in DATA_FILENAMES.items():
        file = data_dir / name
        with file.open(encoding="utf-8-sig", newline="") as handle:
            reader = csv.DictReader(handle)
            if not {"reference", "model_a", "model_b"}.issubset(reader.fieldnames or []):
                raise ValueError(f"{file}: columns reference,model_a,model_b required")
            rows = list(reader)
        if not rows:
            raise ValueError(f"{file}: at least one row required")
        for i, row in enumerate(rows, 1):
            if row["reference"] is None or not row["reference"].strip():
                raise ValueError(f"{file} row {i}: reference must not be empty")
            if row["model_a"] is None or row["model_b"] is None:
                raise ValueError(f"{file} row {i}: both model columns required")
            if task == "classification":
                for col in ("reference", "model_a", "model_b"):
                    if row[col].strip().casefold() not in {"positive", "negative"}:
                        raise ValueError(f"{file} row {i}: classification labels must be Positive/Negative")
        tables[task] = rows
    return tables
```

```python
def binary_classification_stats(reference: list[str], predicted: list[str]) -> dict[str, float]:
    """Treat Positive as the positive class; compute accuracy and P/R/F1."""
    if not reference or len(reference) != len(predicted):
        raise ValueError("Reference and prediction lists must be equally long and nonempty")
    y_true = [str(x).strip().casefold() for x in reference]
    y_pred = [str(x).strip().casefold() for x in predicted]
    if not set(y_true + y_pred).issubset({"positive", "negative"}):
        raise ValueError("Only Positive and Negative class labels are supported")
    tp = sum(t == p == "positive" for t, p in zip(y_true, y_pred))
    tn = sum(t == p == "negative" for t, p in zip(y_true, y_pred))
    fp = sum(t == "negative" and p == "positive" for t, p in zip(y_true, y_pred))
    fn = sum(t == "positive" and p == "negative" for t, p in zip(y_true, y_pred))
    precision = tp / (tp + fp) if tp + fp else 0.0
    recall = tp / (tp + fn) if tp + fn else 0.0
    f1 = 2 * precision * recall / (precision + recall) if precision + recall else 0.0
    return {"accuracy": 100.0 * (tp + tn) / len(y_true),
            "precision": 100.0 * precision, "recall": 100.0 * recall,
            "f1": 100.0 * f1, "tp": tp, "tn": tn, "fp": fp, "fn": fn}


def corpus_edit_rate(rows: list[dict[str, str]], model_column: str, metric: str) -> float:
    """Compute corpus WER or CER: sum edit errors / sum reference units."""
    if metric not in {"WER", "CER"}:
        raise ValueError("metric must be WER or CER")
    errors = units = 0
    for row in rows:
        if metric == "WER":
            score = count_word_errors(row["reference"], row[model_column])
            errors += score["total_errors"]
            units += score["reference_words"]
        else:
            score = count_character_errors(row["reference"], row[model_column])
            errors += score["total_errors"]
            units += score["reference_characters"]
    return 100.0 * errors / units


def optional_meteor(rows: list[dict[str, str]], model_column: str) -> float | None:
    """NLTK METEOR using Topic 6's clearly documented LIMITED offline synonyms."""
    try:
        from topic_06_meteor_score import scorer_wordnet, score_sentence
    except ImportError:
        return None
    offline_synonyms = scorer_wordnet(False)
    return mean(score_sentence(row["reference"], row[model_column], offline_synonyms) for row in rows)

```

```python
def optional_real_bertscore(rows: list[dict[str, str]], model_type: str) -> tuple[float, float]:
    """Use a real pretrained encoder; never replace with fabricated similarities."""
    try:
        from bert_score import score as score_bertscore
    except ImportError as exc:
        raise RuntimeError("Install requirements-topic-07.txt to compute real BERTScore") from exc
    references = [row["reference"] for row in rows]
    candidates = [row["model_a"] for row in rows] + [row["model_b"] for row in rows]
    _, _, f1 = score_bertscore(
        candidates, references + references,
        model_type=model_type, lang="en", idf=False,
        rescale_with_baseline=False, batch_size=4, verbose=False,
    )
    f1_values = f1.tolist()
    n = len(rows)
    return (100.0 * mean(f1_values[:n]), 100.0 * mean(f1_values[n:]))

```

```python
def compare_all(tables: dict[str, list[dict[str, str]]],
                compute_bertscore: bool = False,
                model_type: str = "distilbert-base-uncased") -> list[dict]:
    """Return measured demo scores, with BERTScore marked not computed by default."""
    results = []

    def add(task: str, metric: str, direction: str, a: float | None, b: float | None,
            explanation: str, status: str = "computed") -> None:
        results.append({
            "task": task, "metric": metric, "better": direction,
            "demo_model_a_percent": "" if a is None else f"{a:.2f}",
            "demo_model_b_percent": "" if b is None else f"{b:.2f}",
            "status": status, "method": explanation,
            "data_origin": "INVENTED CLASSROOM EXAMPLES - NOT REAL LLM OUTPUTS",
        })

    qa = tables["question answering"]
    add("Question answering", "Exact Match (raw)", "Higher",
        em_score([r["model_a"] for r in qa], [r["reference"] for r in qa]),
        em_score([r["model_b"] for r in qa], [r["reference"] for r in qa]),
        "Identical characters, no normalization")

    classification = tables["classification"]
    truth = [r["reference"] for r in classification]
    a = binary_classification_stats(truth, [r["model_a"] for r in classification])
    b = binary_classification_stats(truth, [r["model_b"] for r in classification])
    for label, key in (("Accuracy", "accuracy"), ("Precision", "precision"),
                       ("Recall", "recall"), ("F1-score", "f1")):
        add("Classification", label, "Higher", a[key], b[key],
            "Binary class Positive; macro classes not averaged")

    gen = tables["text generation"]
    refs = [r["reference"] for r in gen]
    bleus = []
    rouges = []
    meteors = []
    for model in ("model_a", "model_b"):
        predictions = [row[model] for row in gen]
        bleus.append(corpus_bleu(refs, predictions, max_n=4)["bleu"])
        rouge_by_example = [score_summary(ref, pred) for ref, pred in zip(refs, predictions)]
        rouges.append({key: mean(s[key]["f1"] for s in rouge_by_example)
                       for key in ("rouge_1", "rouge_2", "rouge_l")})
        meteors.append(optional_meteor(gen, model))
    add("Text generation", "BLEU-4", "Higher", *bleus,
        "Unsmoothed, single-reference, equal-weight corpus BLEU-4")
    for label, field in (("ROUGE-1 F1", "rouge_1"), ("ROUGE-2 F1", "rouge_2"),
                         ("ROUGE-L F1", "rouge_l")):
        add("Text generation", label, "Higher", rouges[0][field], rouges[1][field],
            "Unweighted mean of per-sentence F1 scores")
    add("Text generation", "METEOR", "Higher", *meteors,
        "NLTK sentence mean; limited offline synonym list" if all(v is not None for v in meteors)
        else "Requires NLTK; pip install -r requirements.txt",
        status="computed" if all(v is not None for v in meteors) else "not_computed_missing_nltk")
    if compute_bertscore:
        bert_a, bert_b = optional_real_bertscore(gen, model_type)
        add("Text generation", "BERTScore F1", "Higher", bert_a, bert_b,
            f"Actual pretrained encoder: {model_type}; unrescaled F1")
    else:
        add("Text generation", "BERTScore F1", "Higher", None, None,
            "Not computed; run --real-bertscore to download a pretrained encoder",
            status="not_computed_optional_model")

    tx = tables["transcription"]
    for metric in ("WER", "CER"):
        add("Transcription", metric, "Lower",
            corpus_edit_rate(tx, "model_a", metric),
            corpus_edit_rate(tx, "model_b", metric),
            "Corpus total Levenshtein edits / total reference words" if metric == "WER"
            else "Corpus total edits / Unicode reference code points")
    return results
```

```python
def save_csv(path: Path, rows: list[dict]) -> None:
    path.parent.mkdir(parents=True, exist_ok=True)
    with path.open("w", encoding="utf-8", newline="") as handle:
        writer = csv.DictWriter(handle, fieldnames=list(rows[0]))
        writer.writeheader()
        writer.writerows(rows)


def run(data_dir: Path, results_dir: Path, real_bertscore: bool = False,
        model_type: str = "distilbert-base-uncased") -> list[dict]:
    data = read_data(data_dir)
    rows = compare_all(data, real_bertscore, model_type)
    file = results_dir / ("topic_10_comparative_real_bertscore_on_demo_text.csv" if real_bertscore
                          else "topic_10_comparative_demo_results.csv")
    save_csv(file, rows)
    print("Topic 10: Comparative Evaluation of LLM Metrics")
    print("DEMO DATA: Invented predictions; NO actual LLM responses were collected.")
    print("Compare within each task. NEVER rank different tasks by raw percentages.\n")
    print(f"{'Task':<20}{'Metric':<21}{'Model A':>12}{'Model B':>12}{'Direction':>12}")
    print("-" * 77)
    for row in rows:
        a = row["demo_model_a_percent"] + "%" if row["demo_model_a_percent"] else "N/A"
        b = row["demo_model_b_percent"] + "%" if row["demo_model_b_percent"] else "N/A"
        print(f"{row['task']:<20}{row['metric']:<21}{a:>12}{b:>12}{row['better']:>12}")
    print(f"\nSaved: {file.relative_to(ROOT) if file.is_relative_to(ROOT) else file}")
    if not real_bertscore:
        print("BERTScore: N/A until you run --real-bertscore (downloads a model).")
    print("For a real experiment, replace all demo predictions with actual two-LLM outputs.")
    return rows


def main() -> None:
    parser = argparse.ArgumentParser(description=__doc__)
    parser.add_argument("--data-dir", type=Path, default=ROOT / "data",
                        help="Folder containing the four Topic 10 CSV files")
    parser.add_argument("--results-dir", type=Path, default=ROOT / "results")
    parser.add_argument("--real-bertscore", action="store_true",
                        help="Compute actual BERTScore (optional dependencies and model download)")
    parser.add_argument("--model-type", default="distilbert-base-uncased")
    args, _ = parser.parse_known_args()
    try:
        run(args.data_dir, args.results_dir, args.real_bertscore, args.model_type)
    except Exception as exc:
        raise RuntimeError("Topic 10 comparative evaluation failed") from exc


if __name__ == "__main__":
    main()
```
